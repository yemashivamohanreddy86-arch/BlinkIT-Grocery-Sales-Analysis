```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

sales_outliers = df[
    (df["Sales"] < lower_bound) |
    (df["Sales"] > upper_bound)
]

print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))
print("Number of Sales Outliers:", len(sales_outliers))
```

```python
print("Unusual Sales Records:")
display(sales_outliers.head(10))
```

```python
outlier_results = []

for column in columns_to_check:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower) |
        (df[column] > upper)
    ).sum()

    outlier_results.append({
        "Column": column,
        "Lower Bound": round(lower, 2),
        "Upper Bound": round(upper, 2),
        "Outlier Count": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

### Task 3.4: High-Performing vs Low-Performing Products and Outlets

The analysis compared product categories and outlet types
using total sales, average sales, average ratings, and
record counts.

The highest total sales product category was [Category Name],
while the lowest was [Category Name].

The highest total sales outlet type was [Outlet Type],
while the lowest was [Outlet Type].

The results highlight differences in sales contribution
across product categories and outlet types.

Lower total sales may be associated with fewer records,
so average sales per record was also considered when

```python
outlet_improvement = (
    df_clean.groupby("Outlet Type")
    .agg(
        Total_Sales=("Sales", "sum"),
        Average_Sales=("Sales", "mean"),
        Record_Count=("Sales", "count"),
        Average_Rating=("Rating", "mean")
    )
)

low_sales_outlets = outlet_improvement[
    outlet_improvement["Average_Sales"]
    < overall_average_sales
].sort_values("Average_Sales")

print("Outlet Types with Below-Average Sales:")
display(low_sales_outlets.round(2))
```
